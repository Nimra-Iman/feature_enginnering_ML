```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
dataset=pd.read_csv('loan.csv')
dataset.head()
```

# imputation

```python
dataset.isnull().sum()
```

```python
dataset['Credit_History'].unique()
```

```python
# sklearn, fillna
# dataset['Gender'].mode()[0]
dataset['Credit_History']=dataset['Credit_History'].fillna(dataset['Credit_History'].mode()[0])
```

```python
dataset.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
si=SimpleImputer()
si.fit(dataset[['LoanAmount']])
```

```python
si.transform(dataset[['LoanAmount']])
```

```python
dataset['LoanAmount']=pd.DataFrame(si.transform(dataset[['LoanAmount']]), columns=['LoanAmount'])
```

```python
dataset.isnull().sum()
```

#  outliers

```python
# ApplicantIncome	CoapplicantIncome	LoanAmount	Loan_Amount_Term
plt.figure(figsize=(4,3))
sns.boxplot(x='ApplicantIncome', data=dataset)
plt.show()
```

```python
Q1=np.percentile(dataset['ApplicantIncome'], 25)
Q3=np.percentile(dataset['ApplicantIncome'], 75)
IQR=Q3-Q1
mini=Q1-(1.5*IQR)
maxi=Q3+(1.5*IQR)
dataset=dataset[dataset['ApplicantIncome']<=maxi]
dataset.head()
```

```python
_3rd_std_to_right= dataset['LoanAmount'].mean()+ (3*dataset['LoanAmount'].std())
dataset=dataset[dataset['LoanAmount']<_3rd_std_to_right]
dataset.head()
```

```python
dataset['CoapplicantIncome'].isnull().sum()
```

#  duplicates

```python
dataset.duplicated().unique()

# no duplicte rows in dataset
```

# encoding

```python
Property_Area, Education	,Self_Employed, Gender,	Married
```

```python
dataset['Married'].unique()
```

```python
li=dataset[['Education','Self_Employed', 'Gender',	'Married']]
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder(drop='first')
ohe.fit(li)
```

```python
ohe.transform(li).toarray()
```

```python
data=pd.DataFrame(ohe.transform(li).toarray(), columns=['edu_not_graduate','self_employed_yes','gender_male','married'])
data.head()
```

```python
dataset[['Education','Self_Employed', 'Gender',	'Married']]=data[['edu_not_graduate','self_employed_yes','gender_male','married']]
```

```python
dataset.head()
```

```python
dataset['Property_Area'].unique()
```

```python
order=[['Urban', 'Rural', 'Semiurban']]
from sklearn.preprocessing import OrdinalEncoder
oe=OrdinalEncoder(categories=order)
oe.fit(dataset[['Property_Area']])
```

```python
dataset['Property_Area']=pd.DataFrame(oe.transform(dataset[['Property_Area']]))
```

```python
dataset.head()
```

```python
sns.histplot(x='ApplicantIncome', data=dataset, kde=True)
plt.show()
```
