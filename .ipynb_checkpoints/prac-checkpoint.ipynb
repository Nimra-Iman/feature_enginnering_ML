```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib as plt
```

```python
data = pd.read_csv("C:/code-fun/MACHINE LEARNING/2_feature_engineering/loan.csv")
data.head() 
```

```python
data.isnull().sum()
```

```python
data.info()
```

```python
cols = data.columns
for i in cols:
    percentage_missing_value = ((data[i].isnull().sum())/data.shape[0])*100
    print(percentage_missing_value)
```

```python
cate = data.select_dtypes(include = "object").columns
for i in cate:
    data[i]= data[i].fillna(data[i].mode()[0])
```

```python
data.shape
```

```python
numeri = data.select_dtypes(include = "float64").columns
for i in numeri:
    data[i] = data[i].fillna(data[i].mean())
    
```

```python
data.isnull().sum()
```

```python
numeri
```

```python
 # outliers
```

```python
q1 = np.percentile(data['CoapplicantIncome'],25)
q3 = np.percentile(data['CoapplicantIncome'],75)
iqr = q3-q1
lower = q1-(iqr*1.5)
upper = q3+(iqr*1.5)
data = data[data['CoapplicantIncome'] <= upper]
```

```python
q1 = np.percentile(data['LoanAmount'],25)
q3 = np.percentile(data['LoanAmount'],75)
iqr = q3-q1
lower = q1-(iqr*1.5)
upper = q3+(iqr*1.5)
data = data[data['LoanAmount'] <= upper]
```

```python
sns.boxplot(x='CoapplicantIncome', data=data)
```

```python
#duplicates:
data = data.drop_duplicates()
```

```python
#encoding:
```

```python
data.head()
```

```python
Gender	Married  Self_Employed
```

```python
oh = data[['Gender',	'Married',  'Self_Employed']]
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(drop = 'first')
d = ohe.fit_transform(oh).toarray()
encoded_data = pd.DataFrame(d, columns = ["gender_male", "married_yes","self_employeed"])
```

```python
data[['Gender',	'Married',  'Self_Employed']] = encoded_data
```

```python
data.head()
```

```python
data['Property_Area'].unique()
```

```python
order = {"Urban":0, "Rural":1, "Semiurban":2}
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder()
d = oe.fit_transform(data[['Property_Area']])
data['Property_Area'] = pd.DataFrame(d)
```

```python
# feature_scaling
```

```python
da = data[['ApplicantIncome']]
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
d = ss.fit_transform(da)
data['ApplicantIncome'] = pd.DataFrame(d)
```
